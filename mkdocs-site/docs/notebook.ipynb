```python
import pandas as pd
import plotly.express as px
from IPython.display import HTML, display

df = pd.read_csv("data.csv")

fig = px.bar(
    df,
    x="Серия",
    y="Точность",
    text="Точность",
    title="Точность по сериям эксперимента"
)

fig.update_layout(
    yaxis_range=[0, 1],
    xaxis_title="Серия",
    yaxis_title="Точность"
)

display(
    HTML(
        fig.to_html(
            full_html=False,
            include_plotlyjs="cdn"
        )
    )
)
```
